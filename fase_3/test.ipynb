```python

import pickle
import os


pad_naar_pkl = "skeleton/data/phase3_audioData/audiodata_batch_1/anechoic/params.pkl"


try:
    with open(pad_naar_pkl, "rb") as f:
        params = pickle.load(f)
        
    print("Gelukt! Het bestand is geladen.\n")
    print("Wat voor type data is het?:", type(params))
    
    # Als het een dictionary is (wat meestal zo is), printen we de 'sleutels'
    if isinstance(params, dict):
        print("\nSleutels (keys) in dit bestand:")
        for key, value in params.items():
            print(f" - {key}: {type(value)}")
            
            # Als het kleine lijstjes of getallen zijn, printen we ze ook even
            if isinstance(value, (int, float, str, list)) and not (isinstance(value, list) and len(value) > 10):
                 print(f"   Waarde: {value}")
    else:
        print("\nInhoud:")
        print(params)

except FileNotFoundError:
    print(f"Oeps, het bestand kon niet gevonden worden op pad: {pad_naar_pkl}")
    print("Check even of het pad exact klopt!")
```

```python
import numpy as np

# Pas dit pad aan naar het .npz bestand dat je wilt bekijken
pad_naar_npz = "skeleton/data/phase3_audioData/audiodata_batch_1/anechoic/lma_16kHz.npz"

# Gebruik 'with' zodat het bestand na het lezen weer netjes wordt gesloten
with np.load(pad_naar_npz) as data:
    
    # data.files geeft een lijst met de namen van alle arrays in het bestand
    print("Sleutels (arrays) in dit .npz bestand:", data.files)
    print("-" * 30)
    
    # Loop door alle gevonden arrays heen om hun vorm (shape) te bekijken
    for naam in data.files:
        array_inhoud = data[naam]
        print(f"Array '{naam}':")
        print(f" - Datatype: {array_inhoud.dtype}")
        print(f" - Vorm (shape): {array_inhoud.shape}")
        
        # Als het een hele kleine array is, print dan de inhoud
        if array_inhoud.size < 15:
            print(f" - Inhoud: {array_inhoud}")
        print()
```
